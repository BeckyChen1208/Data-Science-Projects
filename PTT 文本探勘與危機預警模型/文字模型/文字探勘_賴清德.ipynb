```python
# =====================================================
# 0. 安裝必要套件（若已安裝可跳過，在終端機或 Colab 執行）
# -----------------------------------------------------
# pip install transformers torch pandas matplotlib ckip-transformers
# =====================================================

import pandas as pd
import torch
import numpy as np
import matplotlib.pyplot as plt
from transformers import AutoTokenizer, AutoModelForSequenceClassification
from ckip_transformers.nlp import CkipWordSegmenter  # ★ CKIP 斷詞

# （可選）避免中文字變成框框或警告
plt.rcParams["font.sans-serif"] = ["Microsoft JhengHei", "PingFang TC", "Noto Sans CJK TC"]
plt.rcParams["axes.unicode_minus"] = False

# =====================================================
# 1. 設定裝置
# =====================================================
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("目前使用裝置：", device)

# =====================================================
# 2. 載入情緒分析模型（HuggingFace 預訓練）
# =====================================================
model_name = "uer/roberta-base-finetuned-jd-binary-chinese"

print("正在載入情緒模型與 tokenizer ...")
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name)
model.to(device)
model.eval()  # 推論模式

# =====================================================
# 2.5 載入 CKIP Transformers 中文斷詞模型
# =====================================================
# device 設定：GPU 用 0，CPU 則用 -1
ckip_device = 0 if torch.cuda.is_available() else -1
print("正在載入 CKIP 斷詞模型 ...")
ws_driver = CkipWordSegmenter(model="bert-base", device=ckip_device)

# =====================================================
# 3. 批次情緒推論函式（沿用你原本的 HuggingFace 寫法）
#    這裡的 texts 會改成「已斷詞後的句子（用空白串起來）」
# =====================================================
def batch_sentiment(texts, batch_size=32, max_len=256):
    """
    輸入：texts = List[str]（已由 CKIP 斷詞，詞與詞之間以空白分隔）
    輸出：scores = List[float]，範圍約在 [-1, 1]
        score > 0：偏正面；score < 0：偏負面
    """
    scores = []

    for i in range(0, len(texts), batch_size):
        batch = texts[i:i + batch_size]

        inputs = tokenizer(
            batch,
            max_length=max_len,
            truncation=True,
            padding=True,
            return_tensors="pt"
        )

        # 丟到 GPU / CPU
        for k in inputs:
            inputs[k] = inputs[k].to(device)

        with torch.no_grad():
            outputs = model(**inputs)

        probs = torch.softmax(outputs.logits, dim=1).cpu().numpy()

        # 假設 logits 的第 0 維是 negative，第 1 維是 positive
        for p_neg, p_pos in probs:
            score = float(p_pos - p_neg)  # 正負差，約在 [-1, 1]
            scores.append(score)

    return scores

# =====================================================
# 4. 主函式：CKIP 斷詞 → 每篇文章一個情緒分數 → 每日平均 → 畫走勢圖
# =====================================================
def run_sentiment_daily_plot(csv_path="賴清德02_完整版_已清理.csv"):
    print(f"正在讀取檔案：{csv_path}")
    df = pd.read_csv(csv_path, encoding="utf-8-sig")
    print("原始資料筆數：", len(df))

    # -------- 4-1. 合併文字欄位（標題 + 內文 + 留言）成一篇 --------
    for col in ["標題", "內文", "留言"]:
        if col not in df.columns:
            df[col] = ""
        df[col] = df[col].fillna("")

    df["full_text"] = (
        df["標題"].astype(str) + "\n" +
        df["內文"].astype(str) + "\n" +
        df["留言"].astype(str)
    )

    # -------- 4-2. 處理日期欄位 --------
    df["date"] = pd.to_datetime(df["時間"], errors="coerce")

    before_drop = len(df)
    df = df.dropna(subset=["date"])
    after_drop = len(df)
    if before_drop != after_drop:
        print(f"有 {before_drop - after_drop} 筆因為日期有問題被丟掉。")
    df = df.sort_values("date")

    # -------- 4-3. 先用 CKIP 對 full_text 做中文斷詞 --------
    print("開始 CKIP 中文斷詞（可能稍微花時間）...")
    texts = df["full_text"].tolist()
    # ws_driver 回傳 List[List[str]]，每一篇是一個詞列表
    ws_list = ws_driver(texts, batch_size=32)

    # 把每一篇的詞用空白串起來，變成一個新欄位
    df["full_text_ckip"] = [" ".join(tokens) for tokens in ws_list]

    # -------- 4-4. 每篇文章（已 CKIP 斷詞）→ 產生一個情緒分數 --------
    print("開始進行情緒分析（較花時間）...")
    df["sentiment"] = batch_sentiment(df["full_text_ckip"].tolist())

    # -------- 4-5. 整體平均情緒（所有筆數一起平均） --------
    overall_mean = df["sentiment"].mean()
    print("\n====== 整體結果 ======")
    print("總筆數：", len(df))
    print("整體平均情緒分數（all data mean）：", round(overall_mean, 4))

    # -------- 匯出結果 CSV ★ ★ ★ --------
    output_path = "賴清德_sentiment_output.csv"
    df.to_csv(output_path, index=False, encoding="utf-8-sig")
    print(f"\n已成功匯出：{output_path}\n")

    # （可選）看一下情緒分布大致狀況
    print("\n情緒分數統計描述（describe）：")
    print(df["sentiment"].describe())

    # （可選）印出幾筆範例，確認模型真的有在評分
    print("\n範例貼文情緒分數（前 5 筆）：")
    print(df[["date", "標題", "sentiment"]].head(5))

    # -------- 4-6. 每一天底下很多篇 → 平均成「每日情緒」 --------
    daily_mean = (
        df.groupby("date")["sentiment"]
          .mean()
          .reset_index()
          .sort_values("date")
    )

    print("\n日期範圍：", daily_mean["date"].min(), "→", daily_mean["date"].max())
    print("\n前 10 天的每日平均情緒（只是 sample，不是只有 10 天）：")
    print(daily_mean.head(10))

    # -------- 4-7. 畫出「每日平均情緒走勢圖」 --------
    plt.figure(figsize=(12, 4))
    plt.plot(
        daily_mean["date"],
        daily_mean["sentiment"],
        linewidth=2
    )
    plt.xlabel("日期")
    plt.ylabel("每日平均情緒")
    plt.title("賴清德 每日平均情緒走勢（CKIP 斷詞 + 預訓練情緒模型）")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

    # 若之後還要拿來做別的圖，可以保留 df / daily_mean
    return df, daily_mean

# =====================================================
# 5. 主程式入口
# =====================================================
if __name__ == "__main__":
    df_lcd, daily_lcd = run_sentiment_daily_plot("賴清德02_完整版_已清理.csv")
```

```python
# ========= 6. 根據情緒區間分組並畫長條圖 =========

import numpy as np
import matplotlib.pyplot as plt

# df_lcd 是你 sentiment 主程式 return 的 DataFrame
df = df_lcd.copy()

# 建立你的四個區間
bins = [-1, -0.5, 0, 0.5, 1]            # 分組邊界
labels = ["-1 ~ -0.5", "-0.5 ~ 0", "0 ~ 0.5", "0.5 ~ 1"]  # 區間名稱

# 用 pandas.cut 做區間分類
df["sentiment_group"] = pd.cut(df["sentiment"],
                               bins=bins,
                               labels=labels,
                               include_lowest=True)

# 計算每區間的數量
count_series = df["sentiment_group"].value_counts().sort_index()

print("情緒區間計數：")
print(count_series)

# ========= 畫長條圖 =========
plt.figure(figsize=(10, 5))
plt.bar(count_series.index, count_series.values, color=['red', 'orange', 'lightgreen', 'green'])
plt.xlabel("情緒分數區間")
plt.ylabel("貼文數量")
plt.title("賴清德 情緒分布長條圖")
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.tight_layout()
plt.show()
```
